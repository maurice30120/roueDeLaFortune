# Roue graphique (Notebook / Colab)

Ce notebook fonctionne:
- en local (depuis le repo)
- sur Google Colab (clone automatique du repo)

Il lance `assign_chores` et affiche une roue **matplotlib** qui tourne et s'arrête sur l'utilisateur sélectionné.


```python
# Setup: trouver/installer le repo puis rendre `chore_manager` importable
import os
import sys
import subprocess
from pathlib import Path

REPO_URL = "https://github.com/maurice30120/roueDeLaFortune.git"
COLAB_PATH = Path("/content/roueDeLaFortune")

def in_colab() -> bool:
    return "google.colab" in sys.modules

def find_repo_root(start: Path) -> Path | None:
    start = start.resolve()
    for candidate in [start, *start.parents]:
        if (candidate / "chore_manager").exists() and (candidate / "app.py").exists():
            return candidate
    return None

if in_colab():
    if not COLAB_PATH.exists():
        subprocess.check_call(["git", "clone", REPO_URL, str(COLAB_PATH)])
    os.chdir(COLAB_PATH)
    subprocess.check_call([sys.executable, "-m", "pip", "-q", "install", "-r", "requirements.txt"])
    subprocess.check_call([sys.executable, "-m", "pip", "-q", "install", "matplotlib"])
else:
    repo = find_repo_root(Path.cwd())
    if repo is None:
        raise RuntimeError("Repo introuvable: place ce notebook dans le repo (ou dans notebooks/) puis relance.")
    os.chdir(repo)

repo_root = Path.cwd().resolve()
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

print("repo_root:", repo_root)
```
